```python
#2.program to check given year is leap year or not
year=int(input("enter a year: "))
if year%400==0 or year%4==0 and year%100!=0 :
   print(f"{year} is a leap year")
else:
  print(f"{year} is not a leap year")
```

```python
#3.example code on match case
day=int(input())
match day:
  case 1:
    print("sunday")
  case 2:
    print("monday")
  case 3:
    print("tuesday")
  case 4:
    print("wednesday")
  case 5:
    print("thursday")
  case 6:
    print("friday")
  case 7:
    print("saturday")
  case _ :
    print("invalid")
```
