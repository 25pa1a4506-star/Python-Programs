```python
#1.List operation
#concatenation
l1=[5,6,9,32,'python']
l2=[1,2,33,46]
print(l1+l2)
```

```python
#repetation
print(l1*2)
```

```python
#indexing
print(l1[1],l2[1])
```

```python
#slicing
print(l1[0:2])
```

```python
#membership operator
l3=[1,2,3,9,32,33,46,11]
if 9 in l3:
  print("yes")
else:
  print("no")
print("end")
```

```python
#iteration
for i in l1:
  print(i)
```

```python
#reverse a list using range
print(l1[ : :-1])
```

```python
#bulit in func
#len()
print(len(l1))
```

```python
#sum
l4=[1,2]
l5=[5,10]
print(sum(l4+l5))
```

```python
#min
print(min(l3))
```

```python
#max
print(max(l5))
```

```python
#for sorting
print(sorted(l3))
```

```python
#for adding one element to list
l2.append("k")
print(l2)
```

```python
#adding 1 list to another list
l2.extend(l3)
print(l2)
```

```python
#inserting an element a particular index
l6=[55,5,5654]
l6.insert(0,1)
print(l6)
```
