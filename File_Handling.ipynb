```python
#file handiling
#program on "w" mode
file = open("py.","w")
data="Hello,World!"
file.write(data)
file.write("\nfile created")
file.close()
```

```python
#"r" mode
with open("py.","r") as f:
  data=f.read()
  print(data)
```

```python
with open("py.","a") as f:
  f.writelines(["\napple","\nname","\nend"])
```

```python
with open("py.","r") as f:
  print(f.readline())
```

```python
with open("py.","r") as f:
  print(f.readlines())
```

```python
with open("py.","r")as f:
  print(f.readline())
  print(f.tell())
  f.seek(1)
  print(f.tell())
  print(f.read())
```

```python
with open("py.","r") as f:
  print(f.readlines())
  print(f.tell())
  print(f.seek(4))
  print(f.tell())
  print(f.read())
```

```python
#program to copy a data ina a file and paste into another file
with open("py.","r") as f:
  data=f.read()
  print("reading done")
  with open("f2.txt","w") as f2:
    f2.write(data)
    print("file copied")
```

```python
#prog to print each line in a file in reverse order
with open("py.","r") as f:
  data=f.readlines()
  for i in data:
    print(i[::-1])
```
